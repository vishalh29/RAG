```python
from dotenv import load_dotenv
load_dotenv()

import os
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain_core.messages import AIMessage,HumanMessage
from pprint import pprint

messages=[AIMessage(content=f"Please tell me how can I help",name="LLMModel")]
messages.append(HumanMessage(content=f"I want to learn coding",name="vishal"))
messages.append(AIMessage(content=f"Which programming language you want to learn",name="LLMModel"))
messages.append(HumanMessage(content=f"tell me about langchain",name="vishal"))

for message in messages:
    message.pretty_print()

```

```python
from langchain_groq import ChatGroq
llm=ChatGroq(model="qwen/qwen3-32b")
result=llm.invoke(messages)
```

```python
result.response_metadata
```

```python
import re

think_match = re.search(r'<think>(.*?)</think>', result.content, re.DOTALL)

if think_match:
    think_content = think_match.group(1).strip()
    response_content = result.content[think_match.end():].strip()
else:
    think_content = ""
    response_content = result.content.strip()
```

```python
print("===== Thinking =====\n")
print(think_content)
```

```python
print("===== Response =====\n")
print(response_content)
```

```python
messages.append(AIMessage(content=response_content, name="LLMModel"))

print("===== Conversation History =====\n")
for message in messages:
    message.pretty_print()
    print()
```

```python
user_input = input("You: ")

messages.append(HumanMessage(content=user_input, name="vishal"))

result = llm.invoke(messages)

think_match = re.search(r'<think>(.*?)</think>', result.content, re.DOTALL)
if think_match:
    clean_response = result.content[think_match.end():].strip()
else:
    clean_response = result.content.strip()

messages.append(AIMessage(content=clean_response, name="LLMModel"))

print(f"\nAI: {clean_response}")
```

```python
print("===== Full Conversation History =====\n")
for message in messages:
    message.pretty_print()
    print()
```
