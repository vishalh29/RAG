```python
from langchain_community.document_loaders import Docx2txtLoader,UnstructuredWordDocumentLoader
```

```python
#Method 1 Using Docx2txtLoader

print("Method 1 Using Docx2txtLoader")

try:
    loader = Docx2txtLoader("data/word_files/proposal.docx")
    documents = loader.load()
    print(f"Number of documents: {len(documents)}")
    print(f"Document content: {documents[0].page_content}")
    print(f"Document metadata: {documents[0].metadata}")

except Exception as e:
    print(f"Error loading document: {e}")

```

```python
# Method 2 Using UnstructuredWordDocumentLoader

print("Method 2 Using UnstructuredWordDocumentLoader")

try:
    loader = UnstructuredWordDocumentLoader("data/word_files/proposal.docx", mode="elements")
    unstructured_documents = loader.load()
    print(f"Number of documents: {len(unstructured_documents)}")
    print(unstructured_documents[0].page_content)
    print(unstructured_documents[0].metadata)

    for i in range(len(unstructured_documents)):
        print(f"Document {i+1} content: {unstructured_documents[i].page_content}")
        print(f"Document {i+1} metadata: {unstructured_documents[i].metadata}")

except Exception as e:
    print(f"Error loading document: {e}")
```
